```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.integrate import solve_ivp
```

```python
def sys(t,y,A,C,B):
  P1 = 1 * np.sin(2*np.pi*0.5*t)
  v1, v2 = y
  d_v1 = v2
  d_v2 = (R*P1 - (1/A)*v2 - v1)/R
  return [d_v1, d_v2]

def sys(t,y,A,C,B):
  P2 = 3 * np.sin(2*np.pi*1*t)
  v1, v2 = y
  d_v1 = v2
  d_v2 = (R*P - (1/A)*v2 - v1)/R
  return [d_v1, d_v2]
```

```python
# Constantes para la simulación (PA: Parametro A,PB: Parametro B,PC:Parametro C)
R = 0.5
P=0.2
  # Experimento 1
P_A = 1
P_B = 3
P_C =0.2
in_s = [0, 1]

  # Experimento 2
E_A = 2.0
E_B = 0.2
E_C = 1
in_d = [0, 0.5]

# Tiempo
start = 0
stop = 20
step = 0.01
time = np.arange(start, stop, step)
t_span = (start, stop)
```

```python
# Simulación para el experimento 1
sol_s = solve_ivp(sys, t_span, in_s, args=(R,P_A,P_B ), t_eval=time)
t_s = sol_s.t      # Tiempo
dv1_s = sol_s.y[0]
dv2_s = sol_s.y[1]

# Simulación para el experimeto 2
sol_d = solve_ivp(sys, t_span, in_d, args=(R,E_A, E_B, P_C), t_eval=time)
t_d = sol_d.t      # Tiempo
dv1_d = sol_d.y[0]
dv2_d = sol_d.y[1]
```

```python
#3.Comparacion entre los dos experimentos
plt.figure(figsize=(10, 6))
plt.title('Comparación del Experimento 1 y Experimento 2')
plt.plot(t_s, dv1_s, label='Experiemnto 1', color='blue', linestyle='dashed')
plt.plot(t_d, dv1_d, label='Experimento 2', color='red', linestyle='dotted')
plt.xlabel('Tiempo[s]')
plt.xlim(start, stop)
plt.ylabel('Posicion de la particula')
plt.legend()
plt.grid(True)
plt.show()
```
