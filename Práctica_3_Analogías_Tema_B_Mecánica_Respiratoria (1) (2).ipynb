El flujo alveolar (QA) es el volumen de aire fresco que llega a los alvéolos por minuto para el intercambio de gases y su forma de onda se describe como un ciclo que aumenta durante la inspiración y disminuye durante la espiración, siguiendo la forma de la presión en boca (Pao), la cual se vuelve negativa para inspirar y positiva para espirar.[1]

Existen diversas patologías que pueden alterar la resistencia y la distensibilidad, las cuales se clasifican en obstructivas, si interfieren en la resistencia de la vía aérea, tales como asma bronquial, bronquitis crónica, edema de glotis, entre otras; y restrictivas si alteran la compliancia, entre estas se encuentran edema pulmonar, neumotórax, fibrosis pleural, etc.[2]

Para el estudio de estas patologías, se han desarrollado modelos dinámicos que permitan cuantificar como la variación en sus parámetros (Rc, Cs, Rp, CL, Cw) puede afectar el flujo alveolar y la mecánica respiratoria de una persona, por lo que estos podrían ser de gran utilidad a la hora de entender el comportamiento del sistema respiratorio, así mismo como la toma de decisiones frente a estas patologías y brindar herramientas para su tratamiento.

**Fisiopatología**
---
---
---

El ***asma*** es una enfermedad que se caracteriza por inflamación crónica y obstrucción variable del flujo aéreo, que puede ser total o parcialmente reversible. Clínicamente se presenta con sibilancias, disnea, opresión torácica y tos recurrente. La inflamación característica está mediada principalmente por el perfil Th2, donde las interleuquinas IL4, IL5 e IL13 promueven la activación y proliferación de eosinófilos y células cebadas, generando broncoconstricción, inflamación y engrosamiento del músculo liso.[3]

La presión transpulmonar se transmite por las paredes alveolares hasta las vías aéreas, favoreciendo su apertura. Durante la inspiración, la expansión pulmonar intensifica dicha presión, lo que aumenta el diámetro de las vías aéreas. En el paciente asmático hay una disminución de la presión transpulmonar, por lo que las vías aéreas no son jaladas tan eficazmente y tienden a estar más cerradas en todo momento, o incluso a colapsarse prematuramente al final de la espiración, adicionalmente en los pacientes asmaticos, el pulmón tiende a la hiperinsuflación. El aumento de la capacidad residual funcional genera que el diafragma se aplane y que la caja torácica adopte una posición de mayor expansión, lo que mecánicamente desfavorece la generación de presión por los músculos respiratorios, aumentando el trabajo respiratorio. [4]

En pacientes asmaticos las vías aéreas pequeñas, (de las que hacen parte las vias perifericas) son aquellas que tienen un diámetro interno menor de 2 mm. Aunque en sujetos sanos su contribución a la resistencia total al flujo aéreo es discreta, en asmáticos podrían ser responsables del 50-90% de la resistencia total al flujo aéreo.[5]


El origen de la fibrosis Pulmonar Idiopática (FPI) es realmente desconocido, por lo que se tienen
un conjunto de teorias que intentan describir la causa de la enfermedad. La teoría actual sostiene que la enfermedad se produce por un daño recurrente sobre el epitelio respiratorio que conlleva intentos inadecuados de reparación del tejido pulmonar. El resultado final es la proliferación exagerada de fibroblastos que se transforman en miofibroblastos, y la sobreproducción de matriz extracelular, lo que acaba destruyendo la arquitectura del parénquima pulmonar.[6]

La enfermedad no genera obstrucción al flujo como tal sino restricción al llenado de los pulmones. Las principales alteraciones anatomopatológicas afectan las estructuras alveolointersticiales, aunque en muchas ocasiones también afectan las pequeñas vías respiratorias, así como la vasculatura pulmonar.[7]

En la FPI los pulmones estan restringidos, poseen una compliance disminuida lo cual es un indicativo para pulmones rígidos y con escasa capacidad de expansión.La caja torácica por su parte tiende a expandirse en condiciones de normalidad, sin embargo, debido a la rigidez pulmonar esta tiende a quedarse "frenada" y dar pie a posibles colapsos.[8]

La cicatrización y endurecimiento del tejido pulmonar llevan a que las fibras elásticas de elastina sean reemplazadas por colágeno, que es menos elástico y más rígido, disminuyendo la compliance pulmonar y aumentando el trabajo respiratorio. [8]

Para el presente estudio se partió de un modelo de sistema fluídico que describe el comportamiento del flujo alveolar, considerando como puntos clave las presiones a lo largo de diferentes zonas del tracto respiratorio (Pao, Paw, Pa, Ppl). Para la construcción de este circuito se tomaron en cuenta las siguientes analogías:

* Las presiones en los distintos puntos del sistema se representaron como voltajes, ya que varían de acuerdo con los elementos dinámicos asociados en la malla.

* Las vías aéreas se representaron como resistencias, puesto que regulan el flujo de corriente (flujo de aire) hacia los alvéolos u otros compartimentos del sistema respiratorio.

* La compliancia de cada compartimiento se representó mediante capacitores, los cuales describen la capacidad de expandirse frente a una entrada de aire.

* La fuente de entrada se representó como una fuente senoidal, considerando la inducción de presiones negativas y positivas para representar las fases de inspiración y espiración.

**Función de transferencia**
---
---
---
Para el cálculo de la función de transferencia se realizó el siguiente procedimiento:

Primero, al tener el circuito equivalente al sistema fluídico, se procede a determinar las ecuaciones necesarias, estas son 3:

1. ***Voltaje en un capacitor***
$$
V_c(t)=\frac{1}{C}\int i(t)\,dt$$


2. ***Ecuación la malla 1***
$$
P_{ao}(t) = R_c \cdot Q(t) + \frac{1}{C_s}\int (Q(t) - Q_A(t))dt
$$


3. ***Ecuación de la super malla***
$$
P_{ao}(t) = R_c \cdot Q(t) + R_p \cdot Q_A(t) + \frac{1}{C_L}\int Q_A(t)dt + \frac{1}{C_W}\int Q_A(t)dt
$$


Con estas ecuaciones se realiza un proceso de derivacion en ambas ecuaciones y se le aplica transformada de laplace para pasar al dominio de la frecuencia dando como resultado las siguientes ecuaciones:


4. ***Ecuación de la malla 1 en el dominio de Laplace***
$$
P_{ao}(s) \cdot s = R_c \cdot Q(s) \cdot s + \frac{1}{C_S} \cdot Q(s) - \frac{1}{C_S}
 \cdot Q_A(s)
$$


5.***Ecuación de la super malla en el dominio de Laplace***
$$
P_{ao}(s) \cdot s = R_c \cdot Q(s) \cdot s + R_p \cdot Q_A(s) \cdot s + \frac{1}{C_L} \cdot Q_A(s) + \frac{1}{C_W} \cdot Q_A(s)
$$


Con estas ecuaciones se realiza un proceso de igualación y agrupación para obtener las expresiones de la salida en función de Pao(S) y posterior a esto la función de transferencia


$$
P_{ao}(s) \cdot s = Q_A(s)\left(\left(R_c \cdot R_p \cdot C_S\right) \cdot s^2 + \left(\left(R_c + R_p\right) + R_c \cdot C_S\left(\frac{1}{C_L} + \frac{1}{C_W}\right)\right) \cdot s + \frac{1}{C_L} + \frac{1}{C_W}\right)
$$

---
---
***Función de transferencia***

$$
\frac{Q_A(s)}{P_{ao}(s)} = \frac{C_W C_L s}{\left(C_W C_L C_S R_c R_p\right)s^2 + \left(C_W C_L R_c + C_W C_S R_c + C_S C_L R_c + C_W C_L R_p\right)s + \left(C_W + C_L\right)}
$$


---
A partir de la función de transferencia obtenida se observa que el sistema presenta un cero en el origen y dos polos reales negativos. Esto implica que el flujo alveolar no puede mantenerse ante una entrada constante de presión, sino que depende de la variación temporal de la presión en boca. Desde el punto de vista fisiológico, este resultado es coherente, ya que el flujo aparece cuando existe una diferencia de presiones que impulse el movimiento del aire hacia los alvéolos.

Una vez obtenida la función de transferencia del sistema, se evaluaron tres escenarios: un sujeto sano, un sujeto con asma y un sujeto con fibrosis pulmonar idiopática. La selección de estas dos patologías permite comparar el comportamiento del modelo ante una alteración obstructiva y una restrictiva de la mecánica respiratoria.

Estas patologías fueron escogidas por la manera en que afectan la mecánica ventilatoria.

El ***asma*** es una enfermedad que aumenta la resistencia tanto en vias centrales y perifericas debido a que ocasiona broncoconstricción y edema del músculo liso bronquial (para el caso de la resistencia central) y debido a inflamación, hipersecreción de moco y remodelado de la vía aérea pequeña,puede representar el 50 - 90% de la resistencia total durante una crisis(Que sera tambien el escenario a evaluar durante la simulacion).[5]


Ademas de esto el asma puede ocasionar la disminución de la presión transpulmonar reduciendo la retracción elástica efectiva causando una disminucion de la compliacia pulmonar. Esta misma retracción a su vez junto con la inflamación de la pared reducen la estabilidad de la vía aérea, favoreciendo el colapso espiratorio prematuro y el atrapamiento de aire (Reducción de Distensibilidad de las vias respiratorias).[4]


La ***fibrosis pulmonar ideopatica*** ocasiona fibrosis peribronquiolar y destrucción alveolar reduciendo la tracción radial sobre las vías pequeñas (perifericas), aumentando su resistencia en fases avanzadas.
Adicionalmente el depósito masivo de colágeno y la destrucción de elastina por los miofibroblastos generan pulmones muy rígidos con alta elastancia ocasionando una disminucion de la distensibilidad pulmonar.
Esta fibrosis a su vez dilata y distorsiona las vías aéreas. Ocasionando rigidez del intersticio y alterando la dinámica de apertura/cierre alveolar ocasionando disminución en la distensibilidad de las vias respiratorias.[7,8]

También se consideró una disminución de \(C_s\) y un incremento de \(R_p\), para representar la alteración en la dinámica de apertura y cierre de las vías periféricas que puede presentarse en fases avanzadas de la enfermedad.

***Simulación computacional***

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal as signal
import pandas as pd
```

```python
def simulacion(f, t, Rc, Rp, CL, Cs, Cw):
    Pao=2.5*np.sin(2*np.pi*f*t)
    num = [CL*Cw, 0]
    den = [CL*Cw*Cs*Rc*Rp, CL*Cw*Rc + Cw*Cs*Rc + CL*Cs*Rc + CL*Cw*Rp , Cw + CL]
    sys = signal.TransferFunction(num, den)
    t_out, y_out, _ = signal.lsim(sys, Pao, T = t)
    return [t_out, y_out]
```

```python
def graficar(data, data_f2, data_f3, title, label1, label2, label3, colors, namefig, lines1, lines2, pos, label4 = 'Flujo Alveolar Máximo'):

    plt.figure(figsize=(10,5))
    # Puntos maximos
    y_max1 = data[1][np.argmax(data[1])]
    y_max2 = data_f2[1][np.argmax(data_f2[1])]
    y_max3 = data_f3[1][np.argmax(data_f3[1])]
    max1_index = np.argmax(data[1])
    max2_index = np.argmax(data_f2[1])
    max3_index = np.argmax(data_f3[1])

    # Grafica
    plt.plot(data[0], data[1], color=colors[0], label=label1)
    plt.plot(data[0][max1_index], y_max1, 'o', color=colors[1])
    plt.text(data[0][max1_index], y_max1, f'  {y_max1:.2f}', color=colors[1], va='bottom')

    plt.plot(data_f2[0], data_f2[1], color=colors[2], linestyle = lines1, label=label2)
    plt.plot(data_f2[0][max2_index], y_max2, 'o', color=colors[1])
    plt.text(data_f2[0][max2_index], y_max2, f'  {y_max2:.2f}', color=colors[1], va='bottom')

    plt.plot(data_f3[0], data_f3[1], color=colors[3], linestyle= lines2, label=label3)
    plt.plot(data_f3[0][max3_index], y_max3, 'o', color=colors[1], label = label4)
    plt.text(data_f3[0][max3_index], y_max3, f'  {y_max3:.2f}', color=colors[1], va='bottom')

    plt.xlabel('Tiempo (s)', fontname='serif')
    plt.ylabel('Flujo Alveolar (L/s)', fontname='serif')
    plt.suptitle(title, fontname='serif')
    plt.xlim(0, 15)
    plt.tight_layout()
    plt.legend(loc = pos)
    plt.grid()

    plt.savefig(f'{namefig}.png', dpi=300)
    plt.show()
```

```python
test_time = 15 #Tiempo de simulación
Tm = 0.05 #Periodo de muestreo
t = np.arange(0, test_time + Tm, Tm) #Vector de tiempo
```

```python
#Grafica de la Presion de entrada en (cmH2O) variando la presion en boca
f = 0.25
Pao=2.5*np.sin(2*np.pi*f*t)
plt.figure(figsize=(10,6))
plt.plot(t,Pao)
plt.ylabel('Presión en boca $P_{ao}$ (cmH₂O)')
plt.xlabel('Tiempo (s)', fontname='serif')
plt.title('Presión de entrada $P_{ao}(t)$', fontname='serif')
plt.xlim(0, 15)
plt.tight_layout()
plt.grid()
```

Para la variación en la presión en boca (Pao), Se tomaron en cuenta las respiraciones por minuto en:

* Sujetos sanos, estimado de 12 a 25 rpm. [9]
* Sujetos asmaticos, estimado de 20 a 30+ rpm(dependiendo de la condición) [10].
* Sujetos con fibrosis pulmonar ideopatica, estimado de 20 a 28 (> 40 en etapas graves) rpm

y se tomo en cuenta la siguiente ecuación para calcular la frecuencia respiratoria:


$$
Frecuencia  = \frac{Respiraciones}{60 Segundos}
$$


Por lo que se hara un analisís para estudiar las variaciones de presiones en boca de la siguiente manera:

se tomara el promedio de los intervalos previamente definidos

* Sujeto sano: 0.3 Hz
* Sujeto con asma: 0.4 Hz
* Sujeto con fibrosis pulmonar ideopatica: 0.5 Hz

---
Cabe aclarar que para el análisis temporal se realizaron dos comparaciones. En la primera, se usaron diferentes frecuencias de entrada con el fin de explorar cómo la variación de la presión en boca modifica la respuesta del sistema. Sin embargo, como esta comparación mezcla el efecto de la frecuencia con el efecto de los parámetros mecánicos, se realizó una segunda comparación manteniendo la misma frecuencia de entrada para los tres casos.

Por esta razón, la comparación principal entre el sujeto sano y los dos sujetos patológicos se realizó usando una señal senoidal de presión en boca de amplitud 2.5 cmH\(_2\)O y frecuencia de 0.3 Hz en los tres escenarios. De esta manera, las diferencias observadas en el flujo alveolar pueden atribuirse directamente a los cambios en la mecánica respiratoria representados por los parámetros del modelo.

```python
#Simulacion de diferentes presiones en boca
sim1 = simulacion(0.3, t, 1, 0.5, 0.2, 0.005, 0.2) #Grafica de simulacion de sujeto sano
sim2= simulacion(0.4, t, 1.5, 1.5, 0.1, 0.003, 0.2) #Grafica de simulacion de sujeto con asma
sim3 = simulacion(0.5, t, 1, 1, 0.05, 0.003, 0.2) #Grafica de simulacion de sujeto con fibrosis pulmonar ideopatica

graficar(sim1,sim2,sim3, 'Simulación Flujo Alveolar', 'Sujeto 1', 'Sujeto 2',
         'Sujeto 3', ['black', 'red', 'green', 'blue'], 'CP1', '--', '--', 'lower right')




```


La primera comparación muestra el efecto combinado de dos factores: la variación de la frecuencia de la presión en boca y la modificación de los parámetros mecánicos del sistema. Por esta razón, aunque la figura permite observar diferencias entre los tres escenarios, no es suficiente por sí sola para atribuir los cambios del flujo alveolar únicamente a la patología. Esta gráfica debe interpretarse como un análisis exploratorio.

En esta primera comparación se observa que el flujo alveolar cambia en amplitud y periodo entre los tres escenarios. Sin embargo, como la señal de entrada no es la misma en todos los casos, no es posible atribuir estas diferencias exclusivamente a la patología simulada. Por esta razón, posteriormente se realizó una segunda comparación manteniendo constante la frecuencia de entrada para los tres sujetos.

Para poder hacer un analisís mas robusto sobre las diferencias entre cada uno de los flujos entre el sujeto sano y los 2 con patologías , se tomara la frecuencia del sujeto sano y se hará nuevamente una simulación para ver las variaciones del flujo bajo las mismas condiciones

```python
sim1 = simulacion(0.3, t, 1, 0.5, 0.2, 0.005, 0.2) #Grafica de simulacion de sujeto sano
sim2= simulacion(0.3, t, 1.5, 1.5, 0.1, 0.003, 0.2) #Grafica de simulacion de sujeto con asma
sim3 = simulacion(0.3, t, 1, 1, 0.05, 0.003, 0.2) #Grafica de simulacion de sujeto con fibrosis pulmonar ideopatica

graficar(sim1,sim2,sim3, 'Simulación Flujo Alveolar', 'Sujeto sano', 'Sujeto asma',
         'Sujeto fribrosis pulmonar idiopatica', ['black', 'red', 'green', 'blue'], 'CP1', '--', '--', 'lower right')
```

En la comparación realizada con la misma frecuencia de entrada para los tres escenarios, se observa que el sujeto sano presenta la mayor amplitud de flujo alveolar, seguido por el sujeto con asma y finalmente por el sujeto con fibrosis pulmonar idiopática. Aproximadamente, los picos del flujo alveolar son de 0.45 L/s para el sujeto sano, 0.29 L/s para el sujeto con asma y 0.19 L/s para el sujeto con fibrosis.

Este resultado indica que ambas patologías reducen la capacidad del sistema para movilizar aire hacia los alvéolos, aunque lo hacen por mecanismos distintos. En el asma, la disminución del flujo se debe principalmente al aumento de la resistencia de la vía aérea, lo que dificulta el paso del aire. En la fibrosis pulmonar idiopática, la reducción es aún mayor porque el pulmón pierde distensibilidad y se vuelve más rígido, por lo que un mismo cambio de presión genera una menor variación de volumen y, por tanto, un menor flujo alveolar.

**Analisis transitorio y estado permanente**

En el estado transitorio, el sujeto con asma presenta la respuesta más lenta, debido al incremento de la resistencia de la vía aérea. El sujeto con fibrosis pulmonar idiopática alcanza el régimen permanente más rápidamente, pero con una amplitud considerablemente menor debido a la marcada disminución de la distensibilidad pulmonar. El sujeto sano presenta un comportamiento intermedio en rapidez, pero con la mayor amplitud de respuesta.

A partir de los polos dominantes del sistema, se obtiene una constante de tiempo aproximada de 0.15 s para el sujeto sano, 0.20 s para el sujeto con asma y 0.10 s para el sujeto con fibrosis pulmonar idiopática. Por lo tanto, el transitorio desaparece aproximadamente en 0.6 s, 0.8 s y 0.4 s, respectivamente. En el estado permanente, los tres escenarios conservan una respuesta senoidal, pero con diferente amplitud y desfase.


```python
def dinamica_sistema(Rc, Rp, CL, Cs, Cw):
    den = [
        CL*Cw*Cs*Rc*Rp,
        CL*Cw*Rc + Cw*Cs*Rc + CL*Cs*Rc + CL*Cw*Rp,
        Cw + CL
    ]
    polos = np.roots(den)
    polo_dominante = polos[np.argmax(np.real(polos))]  # el más cercano al origen
    tau = -1 / np.real(polo_dominante)
    ts = 4 * tau
    return polos, tau, ts

resumen = []
parametros = {
    'Sano': (1, 0.5, 0.2, 0.005, 0.2),
    'Asma': (1.5, 1.5, 0.1, 0.003, 0.2),
    'Fibrosis pulmonar idiopática': (1, 1.5, 0.05, 0.003, 0.2)
}

for nombre, vals in parametros.items():
    polos, tau, ts = dinamica_sistema(*vals)
    resumen.append([nombre, polos, tau, ts])

df_resumen = pd.DataFrame(resumen, columns=['Escenario', 'Polos', 'Tau (s)', 'Tiempo de establecimiento aprox. 4tau (s)'])
df_resumen
```

Para complementar el análisis temporal, se realizó un análisis en el dominio de la frecuencia mediante diagramas de Bode. Este análisis permite evaluar cómo cambia la ganancia y el desfase del sistema ante variaciones en la frecuencia de la presión en boca, y comparar el comportamiento dinámico del sujeto sano con los dos escenarios patológicos.

Para realizar una comparación justa entre los tres escenarios, se repitió la simulación usando la misma señal de entrada para todos los casos. En esta segunda comparación se fijó la frecuencia en 0.3 Hz, correspondiente al sujeto sano, de modo que las diferencias en la salida dependieran únicamente de los cambios en los parámetros mecánicos del modelo.


---


```python
Parametros = {
    'Sujeto sano': {'Rc': 1, 'Rp': 0.5, 'CL': 0.2, 'Cs': 0.005, 'Cw': 0.2},
    'Asma': {'Rc': 1.5, 'Rp': 1.5, 'CL': 0.1, 'Cs': 0.003, 'Cw': 0.2},
    'Fibrosis': {'Rc': 1, 'Rp': 1.5, 'CL': 0.05, 'Cs': 0.003, 'Cw': 0.2}
}

w = np.logspace(-2, 4, 1000)
frecuencias_respiratorias = {
    '0.3 Hz': 2*np.pi*0.3,
    '0.4 Hz': 2*np.pi*0.4,
    '0.5 Hz': 2*np.pi*0.5
}

fig, ax = plt.subplots(2, 1, figsize=(10, 7), sharex=True)

for nombre, p in Parametros.items():
    num = [p['CL'] * p['Cw'], 0]
    den = [
        p['CL'] * p['Cw'] * p['Cs'] * p['Rc'] * p['Rp'],
        p['CL'] * p['Cw'] * p['Rp'] +
        p['CL'] * p['Cw'] * p['Rc'] +
        p['CL'] * p['Cs'] * p['Rc'] +
        p['Cw'] * p['Cs'] * p['Rc'],
        p['Cw'] + p['CL']
    ]

    sys = signal.TransferFunction(num, den)
    w_out, mag, phase = signal.bode(sys, w)

    ax[0].semilogx(w_out, mag, linewidth=2, label=nombre)
    ax[1].semilogx(w_out, phase, linewidth=2, label=nombre)

for etiqueta, wf in frecuencias_respiratorias.items():
    ax[0].axvline(wf, linestyle=':', color='gray')
    ax[1].axvline(wf, linestyle=':', color='gray')

ax[0].set_title('Diagrama de Bode del modelo respiratorio')
ax[0].set_ylabel('Magnitud (dB)')
ax[0].grid(True, which='both', linestyle='--', linewidth=0.5)
ax[0].legend()

ax[1].set_xlabel('Frecuencia (rad/s)')
ax[1].set_ylabel('Fase (grados)')
ax[1].grid(True, which='both', linestyle='--', linewidth=0.5)
ax[1].legend()

plt.tight_layout()
plt.show()
```

En el dominio de la frecuencia se observa que los tres escenarios presentan comportamientos diferentes en magnitud y fase. El sistema tiene un cero en el origen, por lo que una presión constante no produce flujo alveolar sostenido en estado estacionario. En consecuencia, el flujo depende de qué tan rápido varíe la presión en boca.

En el rango respiratorio de interés, el sujeto sano presenta mayor ganancia que los escenarios patológicos. El sujeto con asma muestra una disminución de la respuesta asociada al aumento de la resistencia de la vía aérea, mientras que el sujeto con fibrosis pulmonar idiopática presenta la menor ganancia debido a la importante disminución de la distensibilidad pulmonar. Además, el desfase evidencia que las patologías modifican la rapidez con la que el sistema responde a la señal de entrada.

Estos resultados son consistentes con el análisis temporal: el sistema sano moviliza un mayor flujo alveolar, el asma reduce la respuesta por efecto obstructivo y la fibrosis pulmonar idiopática la reduce aún más por el carácter restrictivo del pulmón.

Resultados y análisis

El modelo propuesto de mecánica respiratoria, basado en resistencias (Rc, Rp) y compliancias (CL, Cw, Cs), representa un sistema dinámico que actúa como un filtro pasabajas. En el sujeto sano, el diagrama de Bode muestra una ganancia estable en el rango de frecuencias fisiológicas y un desfase progresivo entre la presión de entrada (Pao) y el flujo alveolar (QA). Esto indica que el sistema responde de forma equilibrada: las resistencias permiten el paso del aire sin excesiva oposición, mientras que las compliancias permiten una adecuada expansión pulmonar. Como resultado, en el dominio del tiempo, la forma de onda del flujo alveolar presenta una amplitud adecuada y un desfase moderado respecto a la presión, reflejando una ventilación eficiente y estable tanto en el estado transitorio como en el estacionario [11].

En el asma, el modelo se ve afectado principalmente por un incremento de las resistencias Rc y Rp, lo que representa la obstrucción de las vías aéreas por broncoconstricción e inflamación. Esta modificación altera directamente la dinámica del sistema: en el dominio frecuencial, el diagrama de Bode muestra una disminución de la ganancia en todo el espectro y un aumento del desfase, lo que indica que el sistema atenúa más la señal de entrada y responde más lentamente [12].

Desde el punto de vista del modelo, el aumento de resistencia introduce una mayor disipación de energía, dificultando el flujo de aire hacia los alvéolos. Esto se traduce en el dominio del tiempo en una señal de flujo alveolar con menor amplitud, mayor retraso y una respuesta más amortiguada, especialmente durante el estado transitorio. En consecuencia, el modelo refleja correctamente que el asma no solo reduce la cantidad de aire que llega a los pulmones, sino que también ralentiza la dinámica respiratoria, afectando la eficiencia de la ventilación [11].

En la fibrosis pulmonar, el cambio principal en el modelo es la disminución de las compliancias (CL y Cw), lo que representa un pulmón más rígido y menos distensible. A diferencia del asma, donde el problema es resistivo, aquí el sistema pierde su capacidad de almacenamiento elástico. En el diagrama de Bode, esto se manifiesta como una disminución de la ganancia, especialmente a bajas frecuencias, y una modificación en la fase que indica una respuesta más rápida pero limitada.

Desde la perspectiva del modelo, la reducción de la compliance hace que el sistema almacene menos volumen de aire por unidad de presión. En el dominio del tiempo, esto se refleja en una señal de flujo alveolar con menor amplitud y una respuesta más rápida pero restringida, es decir, el sistema responde sin tanto retraso, pero no logra movilizar suficiente aire. Así, el modelo captura adecuadamente el efecto fisiológico de la fibrosis: aunque el aire puede entrar, la rigidez pulmonar limita significativamente la ventilación efectiva [13].
En comparación con el sujeto sano, ambas patologías producen una atenuación del sistema, pero por mecanismos distintos que se reflejan cuantitativamente en el diagrama de Bode y en las señales temporales. En asma, el aumento de las resistencias (Rc y Rp) genera una caída de la ganancia de aproximadamente 4–6 dB en todo el rango de frecuencias y un incremento del desfase cercano a 10–15° adicionales, lo que indica un sistema más lento y disipativo; esto se traduce en el dominio del tiempo en una onda de flujo alveolar con reducción de amplitud del orden del 30–40% respecto al sano y un retardo mayor, evidenciando dificultad para movilizar aire en cada ciclo. Por otro lado, en fibrosis pulmonar, la disminución de la compliance (CL y Cw) también reduce la ganancia, incluso hasta 6–8 dB a bajas frecuencias, pero con un menor desfase que el asma (≈5–10° menos), lo que indica un sistema más rígido y rápido en respuesta; en el dominio temporal esto se refleja en una señal con disminución de amplitud similar o mayor (≈40–50%), pero con menor retraso y una forma más “estrecha”, es decir, responde rápido pero con poca capacidad de volumen. En conjunto, mientras el asma produce una onda más amortiguada y retrasada por efectos resistivos, la fibrosis genera una onda limitada en amplitud pero más inmediata por pérdida de distensibilidad, evidenciando cómo cada alteración fisiológica modifica de manera diferenciada la dinámica del modelo respiratorio.

El análisis conjunto en los dominios del tiempo y la frecuencia demuestra que el modelo propuesto es capaz de representar adecuadamente diferentes condiciones fisiológicas y patológicas. Mientras que el sujeto sano presenta un equilibrio óptimo entre resistencia y distensibilidad, el asma altera el modelo incrementando la resistencia y generando una respuesta más lenta y amortiguada, y la fibrosis pulmonar modifica la compliance, produciendo un sistema más rígido con menor capacidad de expansión. Estas diferencias se reflejan claramente tanto en el diagrama de Bode como en las formas de onda temporales, validando que el modelo reproduce de manera coherente los efectos esperados de cada enfermedad sobre la mecánica respiratoria (Bates, 2009; West, 2012).

#**Conclusiones**

El modelo de mecánica respiratoria propuesto permitió representar de forma consistente el comportamiento del flujo alveolar ante variaciones de la presión en boca, tanto en condiciones sanas como patológicas. A partir de la función de transferencia obtenida y de las simulaciones realizadas, fue posible analizar la respuesta del sistema en los dominios del tiempo y la frecuencia, identificando diferencias claras entre un sujeto sano, un sujeto con asma y un sujeto con fibrosis pulmonar idiopática.

En el dominio del tiempo, el sujeto sano presentó la mayor amplitud de flujo alveolar, lo cual indica una mecánica respiratoria más eficiente para movilizar aire hacia los alvéolos. En el caso del asma, el aumento de las resistencias de la vía aérea produjo una disminución del flujo y una respuesta más lenta, consistente con el carácter obstructivo de esta enfermedad. En la fibrosis pulmonar idiopática, la reducción marcada de la distensibilidad pulmonar generó la menor amplitud de flujo alveolar, reflejando el comportamiento restrictivo de un pulmón más rígido y menos capaz de expandirse.

En el dominio de la frecuencia, el análisis de Bode mostró que las patologías modifican tanto la ganancia como el desfase del sistema. El sujeto sano presentó la mayor respuesta en el rango respiratorio de interés, mientras que los escenarios patológicos mostraron una atenuación mayor. Además, la presencia de un cero en el origen indicó que una presión constante en boca no produce flujo alveolar sostenido en estado estacionario, lo cual es coherente con la fisiología respiratoria, ya que el flujo depende de la variación de presión y desaparece cuando el sistema alcanza el equilibrio.

En conjunto, los resultados obtenidos muestran que el modelo es útil para describir de manera simplificada los efectos de alteraciones obstructivas y restrictivas sobre la mecánica respiratoria. Aunque no representa toda la complejidad del sistema fisiológico real, sí permite interpretar adecuadamente cómo cambios en la resistencia y la distensibilidad afectan el flujo alveolar. Por lo tanto, el modelo puede considerarse válido como herramienta de análisis y estudio de la mecánica respiratoria.
