```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat


def identificar_sistema_LSE(mat_filepath):
    """
    Identifica la respuesta al impulso de un sistema usando mínimos
    cuadrados (LSE) y desconvolución.

    Los datos deben estar en un archivo .mat con las variables:
    - u: entrada (np.array)
    - y: salida ruidosa (np.array)
    - T: periodo de muestreo (float)
    - htrue: respuesta real al impulso (np.array)
    """

    # Cargar datos del archivo .mat
    data = loadmat(mat_filepath)

    u = data['u'].flatten()
    y = data['y'].flatten()
    T = float(data['T'].flatten()[0])
    p_original = int(data['p'].flatten()[0])  # Solo para graficar htrue
    htrue = data['htrue'].flatten()

    # Solicitar p al usuario
    p = int(input("Digite la longitud de la memoria del sistema p >> "))

    N = min(len(u), len(y))
    print(f"\nNúmero de ocurrencias (N) = {N}")
```

```python
# Construcción de la matriz de observación UU
UU = np.zeros((N, p))

for i in range(p):
    UU[i:, i] = u[:N-i]

UU = T * UU

# Mínimos cuadrados
h_est = np.linalg.lstsq(UU, y[:N], rcond=None)[0]

# Error de estimación
e = y[:N] - UU @ h_est

# Tiempo
t = np.arange(N) * T
t_htrue = np.arange(len(htrue)) * T
```
