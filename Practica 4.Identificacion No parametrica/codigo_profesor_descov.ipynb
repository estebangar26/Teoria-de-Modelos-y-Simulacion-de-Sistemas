```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import deconvolve
```

```python
# Crear señal de entrada y sistema (el sistema se supone desconocido)
u = np.array([0.01] + list(range(1, 11)))  # Entrada

h = np.concatenate((
    np.ones(2),
    np.zeros(3),
    2 * np.ones(4),
    np.ones(2)
))  # Sistema real

y = np.convolve(u, h)  # Salida por convolución

# Se añade ruido muy pequeño
ruido = np.random.rand(len(y))
# y = 1e-17 * ruido + y
```

```python
# Visualización
plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1)
plt.stem(u, linefmt='b', markerfmt='bo', basefmt=" ")
plt.grid()
plt.title('Entrada', fontweight='bold')

plt.subplot(2, 3, 2)
plt.stem(h, linefmt='g', markerfmt='go', basefmt=" ")
plt.grid()
plt.title('Sistema Real', fontweight='bold')

plt.subplot(2, 3, 3)
plt.stem(y, linefmt='r', markerfmt='ro', basefmt=" ")
plt.grid()
plt.title('Salida', fontweight='bold')


# Desconvolución: identificar h a partir de u e y
hi, remainder = deconvolve(y, u)

plt.subplot(2, 3, 5)
plt.stem(hi, linefmt='c', markerfmt='co', basefmt=" ")
plt.grid()
plt.title('Sistema Identificado', fontweight='bold')

plt.tight_layout()
plt.show()
```
